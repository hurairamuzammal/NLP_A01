```python
# Chunk 1: DATA LOADING WITH 50-25-25 SPLIT (CHARACTER LEVEL)
import collections
from dataclasses import dataclass

SPECIAL_TOKENS = {
    "pad": "<PAD>",
    "unk": "<UNK>",
    "bos": "<SOS>",
    "eos": "<EOS>",
}

@dataclass
class DataSplits:
    train_pairs: list
    valid_pairs: list
    test_pairs: list


class CharacterTokenizer:
    def __init__(self, special_tokens=None):
        self.special_tokens = special_tokens or SPECIAL_TOKENS
        self._ordered_special_tokens = [
            self.special_tokens["pad"],
            self.special_tokens["unk"],
            self.special_tokens["bos"],
            self.special_tokens["eos"],
        ]
        self.stoi = {}
        self.itos = []
        self.fitted = False

    def fit(self, texts):
        counter = collections.Counter()
        for text in texts:
            counter.update(list(text))
        self.itos = list(self._ordered_special_tokens)
        self.stoi = {token: idx for idx, token in enumerate(self.itos)}
        for char, _ in counter.most_common():
            if char not in self.stoi:
                self.stoi[char] = len(self.itos)
                self.itos.append(char)
        self.fitted = True
        print(f"Tokenizer fitted with vocab size: {len(self.itos)}")

    def encode(self, text, add_bos=False, add_eos=False, max_len=None):
        if not self.fitted:
            raise RuntimeError("Tokenizer must be fitted before encoding.")
        tokens = []
        if add_bos:
            tokens.append(self.bos_id)
        for ch in text:
            tokens.append(self.stoi.get(ch, self.unk_id))
        if add_eos:
            tokens.append(self.eos_id)
        if max_len is not None:
            tokens = tokens[:max_len]
        return tokens

    def decode(self, token_ids, skip_special=True):
        chars = []
        for idx in token_ids:
            if 0 <= idx < len(self.itos):
                token = self.itos[idx]
                if skip_special and token in self._ordered_special_tokens:
                    continue
                chars.append(token)
        return "".join(chars)

    @property
    def vocab_size(self):
        return len(self.itos)

    @property
    def pad_id(self):
        return self.stoi[self.special_tokens["pad"]]

    @property
    def unk_id(self):
        return self.stoi[self.special_tokens["unk"]]

    @property
    def bos_id(self):
        return self.stoi[self.special_tokens["bos"]]

    @property
    def eos_id(self):
        return self.stoi[self.special_tokens["eos"]]


def read_lines(path):
    if not os.path.exists(path):
        raise FileNotFoundError(
            f"File not found: {path}. Update CONFIG paths or set the ROMAN_TEXT_PATH/URDU_TEXT_PATH environment variables."
        )
    with open(path, "r", encoding="utf-8") as f:
        lines = [line.strip() for line in f if line.strip()]
    print(f"Loaded {len(lines)} lines from {path}")
    return lines


def filter_pairs(urdu_lines, roman_lines, config):
    valid_pairs = []
    for urdu, roman in zip(urdu_lines, roman_lines):
        if len(urdu) < config["min_len"] or len(roman) < config["min_len"]:
            continue
        if len(urdu) > config["max_len"] or len(roman) > config["max_len"]:
            continue
        valid_pairs.append((urdu, roman))
    print(f"Valid pairs after filtering: {len(valid_pairs)}")
    return valid_pairs


def split_pairs(pairs, train_ratio=0.5, valid_ratio=0.25):
    random.shuffle(pairs)
    n = len(pairs)
    train_end = int(train_ratio * n)
    valid_end = int((train_ratio + valid_ratio) * n)
    return DataSplits(
        train_pairs=pairs[:train_end],
        valid_pairs=pairs[train_end:valid_end],
        test_pairs=pairs[valid_end:],
    )


class ImprovedDataset(Dataset):
    def __init__(self, pairs, src_tokenizer, tgt_tokenizer, max_len):
        self.pairs = pairs
        self.src_tokenizer = src_tokenizer
        self.tgt_tokenizer = tgt_tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.pairs)

    def __getitem__(self, idx):
        src_text, tgt_text = self.pairs[idx]
        src_ids = self.src_tokenizer.encode(src_text, max_len=self.max_len)
        tgt_core = self.tgt_tokenizer.encode(
            tgt_text, max_len=self.max_len - 2
        )  # reserve space for BOS/EOS
        tgt_input = [self.tgt_tokenizer.bos_id] + tgt_core
        tgt_output = tgt_core + [self.tgt_tokenizer.eos_id]
        return {
            "src": torch.tensor(src_ids, dtype=torch.long),
            "tgt_input": torch.tensor(tgt_input, dtype=torch.long),
            "tgt_output": torch.tensor(tgt_output, dtype=torch.long),
            "src_len": len(src_ids),
            "tgt_len": len(tgt_core) + 1,
        }


def create_collate_fn(src_pad, tgt_pad):
    def collate_fn(batch):
        if len(batch) == 0:
            return None
        srcs = [item["src"] for item in batch]
        tgt_inputs = [item["tgt_input"] for item in batch]
        tgt_outputs = [item["tgt_output"] for item in batch]
        src_lens = torch.tensor([item["src_len"] for item in batch], dtype=torch.long)
        src_padded = pad_sequence(srcs, batch_first=True, padding_value=src_pad)
        tgt_input_padded = pad_sequence(tgt_inputs, batch_first=True, padding_value=tgt_pad)
        tgt_output_padded = pad_sequence(tgt_outputs, batch_first=True, padding_value=tgt_pad)
        return {
            "src": src_padded,
            "tgt_input": tgt_input_padded,
            "tgt_output": tgt_output_padded,
            "src_lens": src_lens,
        }

    return collate_fn


def load_and_prepare_data(config):
    urdu_lines = read_lines(config["urdu_txt"])
    roman_lines = read_lines(config["roman_txt"])
    if len(urdu_lines) != len(roman_lines):
        raise ValueError(
            f"Misaligned data: {len(urdu_lines)} Urdu lines vs {len(roman_lines)} Roman lines"
        )
    valid_pairs = filter_pairs(urdu_lines, roman_lines, config)
    splits = split_pairs(valid_pairs)

    src_tokenizer = CharacterTokenizer()
    tgt_tokenizer = CharacterTokenizer()
    src_tokenizer.fit([src for src, _ in valid_pairs])
    tgt_tokenizer.fit([tgt for _, tgt in valid_pairs])

    collate_fn = create_collate_fn(src_tokenizer.pad_id, tgt_tokenizer.pad_id)

    train_dataset = ImprovedDataset(splits.train_pairs, src_tokenizer, tgt_tokenizer, config["max_len"])
    valid_dataset = ImprovedDataset(splits.valid_pairs, src_tokenizer, tgt_tokenizer, config["max_len"])
    test_dataset = ImprovedDataset(splits.test_pairs, src_tokenizer, tgt_tokenizer, config["max_len"])

    train_loader = DataLoader(
        train_dataset,
        batch_size=config["batch_size"],
        shuffle=True,
        collate_fn=collate_fn,
        pin_memory=True,
        num_workers=config["num_workers"],
    )
    valid_loader = DataLoader(
        valid_dataset,
        batch_size=config["batch_size"],
        shuffle=False,
        collate_fn=collate_fn,
        pin_memory=True,
        num_workers=config["num_workers"],
    )
    test_loader = DataLoader(
        test_dataset,
        batch_size=1,
        shuffle=False,
        collate_fn=collate_fn,
    )

    print(
        "Data splits -> train: {0}, valid: {1}, test: {2}".format(
            len(train_dataset), len(valid_dataset), len(test_dataset)
        )
    )
    print(f"Source vocab size: {src_tokenizer.vocab_size}")
    print(f"Target vocab size: {tgt_tokenizer.vocab_size}")
    print("Sample source tokens:", src_tokenizer.itos[:20])
    print("Sample target tokens:", tgt_tokenizer.itos[:20])

    return {
        "train_loader": train_loader,
        "valid_loader": valid_loader,
        "test_loader": test_loader,
        "train_pairs": splits.train_pairs,
        "valid_pairs": splits.valid_pairs,
        "test_pairs": splits.test_pairs,
        "src_tokenizer": src_tokenizer,
        "tgt_tokenizer": tgt_tokenizer,
    }


DATA = load_and_prepare_data(CONFIG)
print("Character-level data pipeline ready.")
```

```python
# Chunk 2: MODEL ARCHITECTURE (CHARACTER-LEVEL SEQ2SEQ)
import torch.nn.functional as F


class ImprovedEncoder(nn.Module):
    def __init__(self, vocab_size, embed_dim, hidden_dim, n_layers, dropout, pad_id):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=pad_id)
        self.rnn = nn.LSTM(
            embed_dim,
            hidden_dim,
            n_layers,
            batch_first=True,
            bidirectional=True,
            dropout=dropout if n_layers > 1 else 0.0,
        )
        self.dropout = nn.Dropout(dropout)

    def forward(self, src, src_lengths):
        embedded = self.dropout(self.embedding(src))
        packed = pack_padded_sequence(
            embedded, src_lengths.cpu(), batch_first=True, enforce_sorted=False
        )
        packed_outputs, (hidden, cell) = self.rnn(packed)
        outputs, _ = pad_packed_sequence(packed_outputs, batch_first=True)
        return outputs, (hidden, cell)


class ImprovedDecoder(nn.Module):
    def __init__(self, vocab_size, embed_dim, hidden_dim, n_layers, dropout, pad_id):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=pad_id)
        self.rnn = nn.LSTM(
            embed_dim,
            hidden_dim,
            n_layers,
            batch_first=True,
            dropout=dropout if n_layers > 1 else 0.0,
        )
        self.output_projection = nn.Linear(hidden_dim, vocab_size)
        self.dropout = nn.Dropout(dropout)

    def forward(self, input_token, hidden, cell):
        embedded = self.dropout(self.embedding(input_token))
        rnn_output, (new_hidden, new_cell) = self.rnn(embedded, (hidden, cell))
        output = self.output_projection(rnn_output.squeeze(1))
        return output, (new_hidden, new_cell)


class ImprovedSeq2Seq(nn.Module):
    def __init__(self, encoder, decoder, device):
        super().__init__()
        self.encoder = encoder
        self.decoder = decoder
        self.device = device

        encoder_hidden_size = encoder.rnn.hidden_size * 2
        decoder_hidden_size = decoder.rnn.hidden_size
        decoder_layers = decoder.rnn.num_layers

        self.hidden_projection = nn.Linear(encoder_hidden_size, decoder_hidden_size)
        self.cell_projection = nn.Linear(encoder_hidden_size, decoder_hidden_size)
        self.decoder_layers = decoder_layers

    def init_decoder_states(self, encoder_hidden, encoder_cell):
        encoder_layers = encoder_hidden.size(0) // 2

        final_forward_h = encoder_hidden[encoder_layers - 1]
        final_backward_h = encoder_hidden[-1]
        final_forward_c = encoder_cell[encoder_layers - 1]
        final_backward_c = encoder_cell[-1]

        final_h = torch.cat([final_forward_h, final_backward_h], dim=1)
        final_c = torch.cat([final_forward_c, final_backward_c], dim=1)

        projected_h = self.hidden_projection(final_h)
        projected_c = self.cell_projection(final_c)

        decoder_hidden = projected_h.unsqueeze(0).repeat(self.decoder_layers, 1, 1)
        decoder_cell = projected_c.unsqueeze(0).repeat(self.decoder_layers, 1, 1)
        return decoder_hidden, decoder_cell

    def forward(self, src, src_lengths, tgt_input, teacher_forcing_ratio=0.5):
        batch_size = src.size(0)
        tgt_steps = tgt_input.size(1)
        vocab_size = self.decoder.output_projection.out_features

        _, (encoder_hidden, encoder_cell) = self.encoder(src, src_lengths)
        decoder_hidden, decoder_cell = self.init_decoder_states(encoder_hidden, encoder_cell)

        outputs = []
        input_token = tgt_input[:, 0:1]

        for t in range(tgt_steps):
            output, (decoder_hidden, decoder_cell) = self.decoder(
                input_token, decoder_hidden, decoder_cell
            )
            outputs.append(output.unsqueeze(1))

            teacher_force = (t < tgt_steps - 1) and (random.random() < teacher_forcing_ratio)
            if teacher_force:
                input_token = tgt_input[:, t + 1 : t + 2]
            else:
                next_token = output.argmax(dim=-1, keepdim=True)
                input_token = next_token

        outputs = torch.cat(outputs, dim=1)
        return outputs


def create_model(config, data_dict):
    print("\n=== Creating model ===")
    encoder = ImprovedEncoder(
        vocab_size=data_dict["src_tokenizer"].vocab_size,
        embed_dim=config["embed_dim"],
        hidden_dim=config["hidden_size"],
        n_layers=config["encoder_layers"],
        dropout=config["dropout"],
        pad_id=data_dict["src_tokenizer"].pad_id,
    )
    decoder = ImprovedDecoder(
        vocab_size=data_dict["tgt_tokenizer"].vocab_size,
        embed_dim=config["embed_dim"],
        hidden_dim=config["hidden_size"],
        n_layers=config["decoder_layers"],
        dropout=config["dropout"],
        pad_id=data_dict["tgt_tokenizer"].pad_id,
    )
    model = ImprovedSeq2Seq(encoder, decoder, DEVICE).to(DEVICE)

    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"Model '{config['name']}' has {trainable_params:,} trainable parameters")
    return model


MODEL = create_model(CONFIG, DATA)
print("Model ready for training.")
```

# Training the Character-Level Architecture
The model is trained for 50 epochs with teacher forcing decay and cosine learning-rate scheduling.

```python
# Chunk 3: TRAINING LOOP FOR SINGLE ARCHITECTURE
import time
from copy import deepcopy
from pathlib import Path


def build_scheduler(optimizer, config, total_steps):
    warmup_steps = max(1, config["warmup_steps"])
    min_scale = config["min_lr"] / config["learning_rate"]

    def lr_lambda(step):
        step = max(1, step)
        if step <= warmup_steps:
            return step / warmup_steps
        progress = (step - warmup_steps) / max(1, total_steps - warmup_steps)
        if config["lr_scheduler"] == "cosine":
            cosine_component = 0.5 * (1 + math.cos(math.pi * progress))
            return max(min_scale, cosine_component)
        return 1.0

    return optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)


def compute_teacher_forcing_ratio(epoch, config):
    if epoch <= config["teacher_warmup_epochs"]:
        return config["teacher_start"]
    decay_epoch = epoch - config["teacher_warmup_epochs"]
    ratio = config["teacher_end"] + (config["teacher_start"] - config["teacher_end"]) * (
        config["teacher_decay"] ** decay_epoch
    )
    return max(config["teacher_end"], ratio)


def evaluate_model(model, dataloader, criterion, pad_id):
    model.eval()
    total_loss = 0.0
    total_samples = 0
    with torch.no_grad():
        for batch in dataloader:
            if batch is None:
                continue
            src = batch["src"].to(DEVICE)
            src_lens = batch["src_lens"].to(DEVICE)
            tgt_input = batch["tgt_input"].to(DEVICE)
            tgt_output = batch["tgt_output"].to(DEVICE)

            outputs = model(src, src_lens, tgt_input, teacher_forcing_ratio=0.0)
            logits = outputs
            loss = criterion(logits.reshape(-1, logits.size(-1)), tgt_output.reshape(-1))

            total_loss += loss.item() * src.size(0)
            total_samples += src.size(0)
    return total_loss / max(1, total_samples)


def translate_sentence(model, sentence, data_dict, config, max_len=220):
    model.eval()
    src_tokenizer = data_dict["src_tokenizer"]
    tgt_tokenizer = data_dict["tgt_tokenizer"]
    with torch.no_grad():
        src_tokens = src_tokenizer.encode(sentence, max_len=config["max_len"])
        if len(src_tokens) == 0:
            return ""
        src_tensor = torch.tensor([src_tokens], dtype=torch.long, device=DEVICE)
        src_lens = torch.tensor([len(src_tokens)], dtype=torch.long, device=DEVICE)

        _, (encoder_hidden, encoder_cell) = model.encoder(src_tensor, src_lens)
        decoder_hidden, decoder_cell = model.init_decoder_states(encoder_hidden, encoder_cell)

        input_token = torch.tensor([[tgt_tokenizer.bos_id]], dtype=torch.long, device=DEVICE)
        generated = []
        for _ in range(max_len):
            output, (decoder_hidden, decoder_cell) = model.decoder(input_token, decoder_hidden, decoder_cell)
            next_token = output.argmax(dim=-1).item()
            if next_token == tgt_tokenizer.eos_id:
                break
            generated.append(next_token)
            input_token = torch.tensor([[next_token]], dtype=torch.long, device=DEVICE)
        return tgt_tokenizer.decode(generated)


def train_model(model, data_dict, config):
    train_loader = data_dict["train_loader"]
    valid_loader = data_dict["valid_loader"]
    tgt_tokenizer = data_dict["tgt_tokenizer"]

    optimizer = optim.AdamW(
        model.parameters(),
        lr=config["learning_rate"],
        weight_decay=config["weight_decay"],
        betas=(0.9, 0.98),
        eps=1e-9,
    )

    total_steps = max(1, len(train_loader) * config["num_epochs"])
    scheduler = build_scheduler(optimizer, config, total_steps)
    criterion = nn.CrossEntropyLoss(
        ignore_index=tgt_tokenizer.pad_id,
        label_smoothing=config["label_smoothing"],
    )

    best_val_loss = float("inf")
    best_state = None
    history = {
        "train_loss": [],
        "val_loss": [],
        "val_bleu": [],
        "learning_rates": [],
    }

    smooth = SmoothingFunction().method1
    early_stop_counter = 0
    global_step = 0

    for epoch in range(1, config["num_epochs"] + 1):
        start_time = time.time()
        teacher_ratio = compute_teacher_forcing_ratio(epoch, config)

        model.train()
        epoch_loss = 0.0
        batches = 0

        for batch in train_loader:
            if batch is None:
                continue
            src = batch["src"].to(DEVICE)
            src_lens = batch["src_lens"].to(DEVICE)
            tgt_input = batch["tgt_input"].to(DEVICE)
            tgt_output = batch["tgt_output"].to(DEVICE)

            optimizer.zero_grad(set_to_none=True)
            outputs = model(src, src_lens, tgt_input, teacher_forcing_ratio=teacher_ratio)
            logits = outputs
            loss = criterion(logits.reshape(-1, logits.size(-1)), tgt_output.reshape(-1))

            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=config["max_grad_norm"])
            optimizer.step()
            scheduler.step()

            epoch_loss += loss.item()
            batches += 1
            global_step += 1

            if batches % 50 == 0:
                print(f"  Epoch {epoch}, Batch {batches}, Loss: {loss.item():.4f}")

        avg_train_loss = epoch_loss / max(1, batches)
        val_loss = evaluate_model(model, valid_loader, criterion, tgt_tokenizer.pad_id)

        bleu_scores = []
        sample_pairs = random.sample(data_dict["valid_pairs"], min(30, len(data_dict["valid_pairs"]))) if data_dict["valid_pairs"] else []
        for src_text, ref_text in sample_pairs:
            pred_text = translate_sentence(model, src_text, data_dict, config, max_len=config["max_len"])
            if pred_text:
                bleu_scores.append(
                    sentence_bleu([ref_text.split()], pred_text.split(), smoothing_function=smooth)
                )
        avg_bleu = sum(bleu_scores) / len(bleu_scores) if bleu_scores else 0.0
        val_error = max(0.0, 1.0 - avg_bleu)

        current_lr = scheduler.get_last_lr()[0]
        history["train_loss"].append(avg_train_loss)
        history["val_loss"].append(val_loss)
        history["val_bleu"].append(avg_bleu)
        history["learning_rates"].append(current_lr)

        elapsed = time.time() - start_time
        print(f"Epoch {epoch}/{config['num_epochs']} - {elapsed:.1f}s")
        print(
            f"  Train Loss: {avg_train_loss:.4f} | Val Loss: {val_loss:.4f} | Val BLEU: {avg_bleu:.4f}"
        )
        print(f"  Validation Error (1-BLEU): {val_error:.4f}")
        print(f"  Teacher Forcing: {teacher_ratio:.3f} | Learning Rate: {current_lr:.2e}")

        example_pair = sample_pairs[0] if sample_pairs else (data_dict["valid_pairs"][0] if data_dict["valid_pairs"] else data_dict["train_pairs"][0])
        example_src, example_ref = example_pair
        example_pred = translate_sentence(model, example_src, data_dict, config, max_len=config["max_len"])
        print("  Sample translation check:")
        print(f"    SRC: {example_src}")
        print(f"    REF: {example_ref}")
        print(f"    PRED: {example_pred}")

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_state = deepcopy(model.state_dict())
            early_stop_counter = 0
            print("  -> New best model!")
        else:
            early_stop_counter += 1
            if early_stop_counter >= config["patience"]:
                print(f"Early stopping triggered at epoch {epoch}")
                break
        print("-" * 60)

    artifact_path = Path(config["artifact_path"])
    artifact_path.parent.mkdir(parents=True, exist_ok=True)

    if best_state is not None:
        model.load_state_dict(best_state)
        cpu_state = {k: v.detach().cpu() for k, v in best_state.items()}
        payload = {
            "config": config,
            "model_state_dict": cpu_state,
            "history": history,
            "src_tokenizer": {
                "itos": data_dict["src_tokenizer"].itos,
                "special_tokens": data_dict["src_tokenizer"].special_tokens,
            },
            "tgt_tokenizer": {
                "itos": data_dict["tgt_tokenizer"].itos,
                "special_tokens": data_dict["tgt_tokenizer"].special_tokens,
            },
        }
        torch.save(payload, artifact_path)
        print(f"Best model saved to '{artifact_path.as_posix()}'")
    else:
        print("Training finished without improvement; no artifact saved.")

    return {
        "history": history,
        "best_val_loss": best_val_loss,
        "model": model,
        "artifact_path": artifact_path.as_posix(),
    }


TRAINING_RESULTS = train_model(MODEL, DATA, CONFIG)
print("Training completed.")
```

# Testing the Character-Level Model

```python
# Chunk 4: EVALUATION OF THE CHARACTER-LEVEL MODEL
import editdistance
import matplotlib.pyplot as plt
import numpy as np


def evaluate_on_test(model, data_dict, config):
    model.eval()
    tgt_tokenizer = data_dict["tgt_tokenizer"]
    smooth = SmoothingFunction().method1

    bleu_scores = []
    cer_scores = []
    detailed_examples = []

    for idx, (src_text, ref_text) in enumerate(data_dict["test_pairs"]):
        pred_text = translate_sentence(model, src_text, data_dict, config, max_len=config["max_len"])
        if pred_text:
            bleu = sentence_bleu([ref_text.split()], pred_text.split(), smoothing_function=smooth)
            cer = editdistance.eval(list(ref_text), list(pred_text)) / max(1, len(ref_text))
            bleu_scores.append(bleu)
            cer_scores.append(cer)
        detailed_examples.append((src_text, ref_text, pred_text))
    avg_bleu = sum(bleu_scores) / len(bleu_scores) if bleu_scores else 0.0
    avg_cer = sum(cer_scores) / len(cer_scores) if cer_scores else 1.0
    return {
        "bleu_scores": bleu_scores,
        "cer_scores": cer_scores,
        "avg_bleu": avg_bleu,
        "avg_cer": avg_cer,
        "examples": detailed_examples,
    }


print("Evaluating the trained model on held-out data...\n")
SAMPLE_SENTENCES = [
    "زندگی تجھ پہ اب الزام کوئی کیا رکھے",
    "عاشقی میں میرؔ جیسے خواب مت دیکھا کرو",
    "اب کے ہم بچھڑے تو شاید کبھی خوابوں میں ملیں",
    "دل میں کیا ہے محبت کے سوا اور کیا ہے",
    "یہ دنیا اک تماشا ہے جہاں ہر شخص اداکار",
]

for i, sentence in enumerate(SAMPLE_SENTENCES, 1):
    translation = translate_sentence(MODEL, sentence, DATA, CONFIG, max_len=CONFIG["max_len"])
    print(f"{i}. Urdu:  {sentence}")
    print(f"   Roman: {translation}")
    print()

TEST_METRICS = evaluate_on_test(MODEL, DATA, CONFIG)
print("Test set results:")
print(f"  Average BLEU: {TEST_METRICS['avg_bleu']:.4f}")
print(f"  Average CER:  {TEST_METRICS['avg_cer']:.4f}")
print(f"  Evaluated examples: {len(TEST_METRICS['examples'])}")

print("\nFirst five detailed examples:")
for idx, (src, ref, pred) in enumerate(TEST_METRICS["examples"][:5], 1):
    print(f"Example {idx}:")
    print(f"  SRC:  {src}")
    print(f"  REF:  {ref}")
    print(f"  PRED: {pred}")
    if pred:
        bleu = sentence_bleu([ref.split()], pred.split(), smoothing_function=SmoothingFunction().method1)
        cer = editdistance.eval(list(ref), list(pred)) / max(1, len(ref))
        print(f"  BLEU: {bleu:.4f}, CER: {cer:.4f}")
    print()

# Visualization of training history
history = TRAINING_RESULTS["history"]
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("Training History", fontsize=16, fontweight="bold")

epochs = range(1, len(history["train_loss"]) + 1)
axes[0].plot(epochs, history["train_loss"], label="Train Loss", linewidth=2)
axes[0].plot(epochs, history["val_loss"], label="Val Loss", linewidth=2)
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Loss")
axes[0].set_title("Loss Curves")
axes[0].legend()
axes[0].grid(True, alpha=0.3)

epochs_bleu = range(1, len(history["val_bleu"]) + 1)
axes[1].plot(epochs_bleu, history["val_bleu"], color="#4ECDC4", linewidth=2)
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("BLEU")
axes[1].set_title("Validation BLEU")
axes[1].grid(True, alpha=0.3)

axes[2].plot(epochs, history["learning_rates"], color="#FF6B6B", linewidth=2)
axes[2].set_xlabel("Epoch")
axes[2].set_ylabel("Learning Rate")
axes[2].set_title("Learning Rate Schedule")
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("char_seq2seq_training_history.png", dpi=300, bbox_inches="tight")
plt.show()

print("Visualization saved as 'char_seq2seq_training_history.png'")
```
